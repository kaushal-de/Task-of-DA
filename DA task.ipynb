```python
df["Neighbourhood"].value_counts().head(10)
```

```python
trend = pd.crosstab(df["Neighbourhood"], df["No-show"])

trend.sort_values(by="Yes", ascending=False).head(10)
```

```python
df["No-show"].value_counts().plot(kind='bar')

plt.title("Appointment Status")
plt.xlabel("Status")
plt.ylabel("Count")

plt.show()
```

```python
df["No-show"].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Appointment Status Percentage")
plt.ylabel("")

plt.show()
```

```python
df["Neighbourhood"].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Neighbourhoods")
plt.xlabel("Neighbourhood")
plt.ylabel("Appointments")

plt.xticks(rotation=45)

plt.show()
```
